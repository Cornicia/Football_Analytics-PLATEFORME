# Étape 1 : Exploration et Extraction des Données de Football
Ce notebook sert de premier point d'entrée pour tester nos outils de récupération de données.
Nous allons utiliser `soccerdata` (qui permet de requêter des sites comme FBref) et `statsbombpy` (qui donne accès aux données ouvertes de StatsBomb).

```python
import pandas as pd
import soccerdata as sd
from statsbombpy import sb

# Affichage complet dans pandas pour mieux voir les tableaux
pd.set_option('display.max_columns', None)

print("Bibliothèques importées avec succès !")
```

## 1. Récupération des classements et des matchs avec FBref (via SoccerData)
FBref est l'une des meilleures sources pour les statistiques générales (xG, tirs, possession...).
Récupérons les données de la Premier League pour la saison 2023-2024.

```python
try:
    print("Connexion à FBref en cours (cela peut prendre quelques secondes)...")
    # On initialise l'objet FBref pour la Premier League anglaise (ENG-Premier League) saison 23/24
    fbref = sd.FBref(leagues="ENG-Premier League", seasons="2324")
    
    # Récupérer le calendrier complet et les résultats
    schedule = fbref.read_schedule()
    
    print("\n--- Les 5 premiers matchs de la saison ---")
    print(schedule.head())
    
except Exception as e:
    print(f"Erreur lors de la récupération avec FBref : {e}")
```

## 2. Découverte des données détaillées avec StatsBomb
StatsBomb fournit des données événementielles ultra-précises (chaque passe, chaque tir avec ses coordonnées X, Y sur le terrain).
Regardons quelles compétitions sont disponibles gratuitement.

```python
try:
    print("\nRécupération des compétitions disponibles sur StatsBomb...")
    competitions = sb.competitions()
    
    # Filtrer par exemple pour trouver la Coupe du Monde ou la Ligue des Champions
    print("\n--- Quelques compétitions disponibles ---")
    print(competitions[['competition_id', 'country_name', 'competition_name', 'season_name']].head(10))
    
except Exception as e:
    print(f"Erreur lors de la récupération avec StatsBomb : {e}")
```
